# Python OOP Assignment — Inheritance Examples

```python
# 1) Single Inheritance: Person -> Student
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no

st = Student("Prachi", 15)
print(st.name)
print(st.roll_no)


# TODO:
# 1. Create object of Student.
# 2. Print name.
# 3. Print roll_no.
```

```python
# 2) Single Inheritance: Animal -> Dog
class Animal:
    def sound(self):
        print("Some sound")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

dg = Dog()
dg.sound()
dg.bark()

# TODO:
# 1. Create Dog object.
# 2. Call sound() from parent class.
# 3. Call bark() from child class.
```

```python
# 3) Constructor Inheritance: Vehicle -> Car
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

cr = Car("Samsung", "GalaxyS24")
print(cr.brand)
print(cr.model)


# TODO:
# 1. Create Car object.
# 2. Print brand.
# 3. Print model.
```

```python
# 4) Multilevel Inheritance: A -> B -> C
class A:
    def showA(self):
        print("This is A")

class B(A):
    def showB(self):
        print("This is B")

class C(B):
    def showC(self):
        print("This is C")

cl = C()
cl.showA()
cl.showB()
cl.showC()


# TODO:
# 1. Create C object.
# 2. Call showA(), showB(), showC().
```

```python
# 5) Multilevel Inheritance with methods: LivingThing -> Animal -> Dog
class LivingThing:
    def living(self):
        print("I am alive")

class Animal(LivingThing):
    def eat(self):
        print("Eating…")

class Dog(Animal):
    def bark(self):
        print("Barking…")

dg = Dog()
dg.living()
dg.eat()
dg.bark()

# TODO:
# 1. Create Dog object.
# 2. Call living(), eat(), bark().
```

```python
# 6) Multiple Inheritance: Father + Mother -> Child
class Father:
    def father_method(self):
        print("Father method")

class Mother:
    def mother_method(self):
        print("Mother method")

class Child(Father, Mother):
    def child_method(self):
        print("Child method")

cd = Child()
cd.father_method()
cd.mother_method()
cd.child_method()

# TODO:
# 1. Create Child object.
# 2. Call father_method(), mother_method(), child_method().
```

```python
# 7) Constructor Inheritance: Product -> Laptop
class Product:
    def __init__(self, name):
        self.name = name

class Laptop(Product):
    def __init__(self, name, ram):
        super().__init__(name)
        self.ram = ram

lp = Laptop("DellXPS", "16GB")
print(lp.name)
print(lp.ram)

# TODO:
# 1. Create Laptop object.
# 2. Print name.
# 3. Print ram.
```

```python
# 8) Multilevel with __init__: Country -> State -> City
class Country:
    def __init__(self, cname):
        self.cname = cname

class State(Country):
    def __init__(self, cname, sname):
        super().__init__(cname)
        self.sname = sname

class City(State):
    def __init__(self, cname, sname, cityname):
        super().__init__(cname, sname)
        self.cityname = cityname

cty = City("India", "Maharashtra", "Kolhapur")
print(cty.cname)
print(cty.sname)
print(cty.cityname)


# TODO:
# 1. Create City object.
# 2. Print cname, sname, cityname.
```

```python
# 9) Multiple Inheritance with methods: Teacher + Player -> Student
class Teacher:
    def teach(self):
        print("Teaching…")

class Player:
    def play(self):
        print("Playing…")

class Student(Teacher, Player):
    def study(self):
        print("Studying…")

st = Student()
st.teach()
st.play()
st.study()

# TODO:
# 1. Create Student object.
# 2. Call teach(), play(), study().
```

```python
# 10) Constructor Inheritance: Shape -> Rectangle
class Shape:
    def __init__(self, color):
        self.color = color

class Rectangle(Shape):
    def __init__(self, color, length, width):
        super().__init__(color)
        self.length = length
        self.width = width

rect = Rectangle("Red", "50cm", "400m")
print(rect.color)
print(rect.length)
print(rect.width)

# TODO:
# 1. Create Rectangle object.
# 2. Print color, length, width.
```

```python
# 11) Class And Function Design Question

orders = [
    {
        "id": 1,
        "student": {"id": "S001", "name": "Aarav"},
        "items": [
            {"name": "Sandwich", "price": 50.0, "qty": 2},
            {"name": "Juice", "price": 30.0, "qty": 1},
        ],
        "payment_method": "Cash"
    },
    {
        "id": 2,
        "student": {"id": "S002", "name": "Neha"},
        "items": [
            {"name": "Pasta", "price": 80.0, "qty": 1},
            {"name": "Juice", "price": 30.0, "qty": 2},
        ],
        "payment_method": "Card"
    },
    {
        "id": 3,
        "student": {"id": "S001", "name": "Aarav"},
        "items": [
            {"name": "Sandwich", "price": 50.0, "qty": 1},
            {"name": "Cookie", "price": 20.0, "qty": 3},
        ],
        "payment_method": "Cash"
    }
]

class Order:
    def __init__(self, o):
        self.id = o ["student"] ["id"]
        self.name = o ["student"] ["name"] 
        self.items = o ["items"]

    def process(self):

        subtotal = sum(i["price"] * i["qty"] for i in self.items)
        print(self.id, self.name, "subtotal:", subtotal)

        discount_price = subtotal * 0.95 if subtotal >= 150 else subtotal
        print(self.id, self.name, "Discount is:", discount_price)

        tax = discount_price * 0.05
        print(self.id, self.name, "subtotal:", subtotal, "Tax is :", tax)

        return discount_price + tax

class OrderManager:
    def __init__(self, orders):
        self.orders = orders

    def revenue(self):
        return sum(Order(o).process() for o in self.orders)
    

print("Total revenue:", OrderManager(orders).revenue())


# For each order add key "subtotal"
# Subtotal = sum of price * qty for all items.
# print "student_id", "student_name" and "subtotal"

# Add more key "discount_price"
# Discount:
# If subtotal ≥ ₹150, apply 5% discount.
# Else, Original subtotal.
# print "student_id", "student_name" and "discount_price"

# Total tax applied  "tax"
# Tax: Apply 5% tax on the amount after discount.
# example subtotal is 100 then 5 is tax
#  print "student_id", "student_name" and "subtotal", "tax"

# Total revenue for the day (sum of final totals).
# print revenue
```

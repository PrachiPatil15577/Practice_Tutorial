**Q22.** Explain (in a comment) why `convert_temperature()` should be an instance method if it depends on an object’s `scale` (C or F). Optionally, show a small example.

```python
class Temperature:
    def __init__(self, value, scale):
        self.value = value
        self.scale = scale

    def convert_temperature(self):
        if self.scale == "c":
            return (self.value * 9 /5) + 32
        else:
            return (self.value - 32) * 5 / 9
         
t1 = Temperature(70, "C")
t2 = Temperature(80, "F")

print(t1.convert_temperature())
print(t2.convert_temperature())
```

**Q24.** Make a class `Date` with class method `from_string(cls, s)` to parse `"DD-MM-YYYY"` and return an object.

```python
# TODO: Write your solution for Q24 below

class Date:
    #date = "05-12-2025"
    def __init__(self, d, m, y):
        self.day = d
        self.month = m
        self.year = y

    #def from_string(cls):
        #print(cls.date) 
    @classmethod
    def from_string(cls, s):
        d, m, y = s.split("-")
        return cls(int(d), int(m), int(y))
    
obj = Date.from_string("05-12-2025")
print(obj.day, obj.month, obj.year)

```

**Q25.** Create a class `AppConfig` with class variable `DEFAULT_TIMEOUT` and class method to change it. Show that new instances use updated default.

```python
# TODO: Write your solution for Q25 below

class AppConfig:

    DEFAULT_TIMEOUT = 20

    def __init__(self, timeout=None):
        self.timeout = timeout if timeout is not None else AppConfig.DEFAULT_TIMEOUT

    @classmethod
    def set_default_timeout(cls, new_timeout):
        cls.DEFAULT_TIMEOUT = new_timeout

app1 = AppConfig()
print("App1:", app1.timeout)

AppConfig.set_default_timeout(40)

app2 = AppConfig()
print("App2:", app2.timeout)

print("after change:", app1.timeout)
```

## G. Decorators (27–30)


**Q27.** Write a simple decorator that prints `"Function started"` before running the function. Test it on a small function.

```python
# TODO: Write your solution for Q27 below

def My_decorator(fun):
    def demo():
        print("Function started")
        fun()
    return  demo

@My_decorator
def test():
    print("welcome to world!!")
test()
```

**Q28.** Create a decorator that prints `"Function ended"` after running the function. Test it.

```python
# TODO: Write your solution for Q28 below

def My_decorator(fun):
    def demo():
        print("Fuction ended")
        fun()
    return demo

@My_decorator
def sample():
    print("ended!!")

sample()
```

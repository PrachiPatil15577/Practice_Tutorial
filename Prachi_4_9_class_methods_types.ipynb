## Q1: Create a class called Student

Requirements:
- Class variable: `school_name = "ABC International School"`
- Constructor: takes `name` and `marks`
- Instance method: `display_details()`
- Class method: `change_school()`
- Static method: `is_pass(marks)` returns "Pass" if marks ≥ 35 else "Fail"


```python
class Student:
    
    school_name = "ABC International School"

    def __init__(self):
        self.name = "Prachi"
        self.marks = 30

    def display_details(self):
        print("name:", self.name)
        print("marks:", self.marks)
        print("school", Student().school_name)

    def change_school(cls):
        Student.school_name = "Xyz"
        print(Student().school_name)

    @staticmethod
    def is_pass(marks):
        if marks >= 35:
            print("pass")
        else:
            return "fail"
    

Student().display_details()
Student().change_school()
Student.is_pass(34)

```

## Q2: Create a class Calculator

Requirements:
- Class variable: `count_operations = 0`
- Instance method: `add(a, b)`
- Class method: `increment_count()`
- Static method: `info()`


```python
class Calculator:
    count_operations = 0

    @classmethod
    def increment_count(cls):
        cls.count_operations += 1

    def add(self, a, b):
        res = a + b
        Calculator.increment_count()
        return res

    # Static method
    @staticmethod
    def info():
        return "This is a Calculator class with add operation."

c = Calculator()

print(c.add(5, 7))           
print(Calculator.count_operations)    
#print(Calculator.info())    
print(c.add(10, 9))
print(Calculator.count_operations)
```

## Q3: Write a class BankAccount

Requirements:
- Class variable: `bank_name = "SBI"`
- Constructor: `name`, `balance`
- Instance method: `deposit(amount)`
- Class method: `change_bank(new_name)`
- Static method: `validate_amount(amount)`


```python
class BankAccount:
    bank_name = "SBI"

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        if BankAccount.validate_amount(amount):
            self.balance += amount
            print(f"deposit {amount} new balance {self.balance}")
        else:
            print("Invalidd")
            
    @classmethod
    def change_bank(cls, new_name):
        cls.bank_name = new_name
        print(f"Bank name changed to {cls.bank_name}")

    @staticmethod
    def validate_amount(amount):
        return amount > 0
    

ba = BankAccount("Prachi", 50000)
ba.deposit(5000)

BankAccount.change_bank("HDFS")
```
